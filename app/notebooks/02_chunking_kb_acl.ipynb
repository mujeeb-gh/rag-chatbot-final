```python
import pandas as pd
```

```python
acl_df = pd.read_csv('cleaned_kb_acl.csv')
acl_df
```

### I'm not meant to have NaN values, so I'll do some extra preprocessing here

```python
acl_df.dropna(inplace=True)
acl_df
```

```python
acl_df.info()
```

### Put both features in lists

```python
titles, texts, acl_ids, urls, years, authors = list(acl_df['title']), list(acl_df['text']), list(acl_df['acl_id']), list(acl_df['url']), list(acl_df['year']), list(acl_df['author'])
```

```python
print(titles[20000])
```

### Now, to chunk into a 100 word splits

```python
print(f"Before splitting, {len(titles)} ACL Corpus articles. \n")

passage_titles = []
passages = []
passage_acl_ids = []
passage_urls = []
passage_years = []
passage_authors = []

# For each text and its title...
for i in range(len(titles)):

    title = titles[i]
    text = texts[i]
    acl_id = acl_ids[i]
    url = urls[i]
    year = years[i]
    author = authors[i]

    # Skip over any without contents
    if len(text) == 0:
        print("Skipping empty text:", text)
        continue

    # Split text on whitespace
    words = text.split()

    # Loop over the words, incrementing by 100
    for i in range(0, len(words), 100):

        # Select the next 100 words
        chunk_words = words[i : i + 100]

        # Recombine thw ords into a passage by joining with whitespace
        chunk = " ".join(chunk_words)

        # Remove any trailinig whitespace
        chunk = chunk.strip()

        # To avoid possible edge cases, skip any empty chunks.
        if len(chunk) == 0:
            continue

        # Store each passage chunk for each title
        passage_titles.append(title)
        passages.append(chunk)
        passage_acl_ids.append(acl_id)
        passage_urls.append(url)
        passage_years.append(year)
        passage_authors.append(author)

print(' Done.\n')

chunked_corpus = {'title': passage_titles, 'text': passages, 'acl_id': passage_acl_ids, 'url': passage_urls, 'year': passage_years, 'author': passage_authors}

print(f"After splitting, {len(chunked_corpus['title'])} '100-word chunk' passages.")
```

### Save the chunked corpus to disk

```python
chunked_df = pd.DataFrame(chunked_corpus)

chunked_df
```

```python
chunked_df.to_csv("chunked_kb_acl.csv", index=False)
```
