```python
from sklearn.model_selection import train_test_split
import json

def load_dataset(file_path):
    with open(file_path, 'r') as f:
        dataset = json.load(f)
    return dataset

def save_dataset(file_path, dataset):
    with open(file_path, 'w') as f:
        json.dump(dataset, f)
    return None
```

```python
DATASET = "/kaggle/input/acl-question-context-dataset-1500/dataset-1500.json"
TRAIN_DATASET_PATH = "/kaggle/working/train_dataset.json"
TEST_DATASET_PATH = "/kaggle/working/test_dataset.json"
VAL_DATASET_PATH = "/kaggle/working/val_dataset.json"
```

```python
dataset = load_dataset(DATASET)
```

```python
import json

def split_data(training_file):
    with open(training_file, 'r') as file:
        data = json.load(file)
    
    total_queries = len(data['queries'])
    total_corpus = len(data['corpus'])
    
    # Calculate the split sizes
    train_size = int(total_queries * 0.7)
    val_size = int(total_queries * 0.15)
    test_size = total_queries - train_size - val_size
    
    # Split queries
    queries = list(data['queries'].items())
    train_queries = dict(queries[:train_size])
    val_queries = dict(queries[train_size:train_size+val_size])
    test_queries = dict(queries[train_size+val_size:])
    
    # Split corpus
    corpus = list(data['corpus'].items())
    train_corpus = dict(corpus[:train_size])
    val_corpus = dict(corpus[train_size:train_size+val_size])
    test_corpus = dict(corpus[train_size+val_size:])
    
    # Split relevant_docs
    relevant_docs = data['relevant_docs']
    train_relevant_docs = {k: v for k, v in relevant_docs.items() if k in train_queries}
    val_relevant_docs = {k: v for k, v in relevant_docs.items() if k in val_queries}
    test_relevant_docs = {k: v for k, v in relevant_docs.items() if k in test_queries}
    
    # Save split data into separate JSON files
    with open(TRAIN_DATASET_PATH, 'w') as file:
        json.dump({'queries': train_queries, 'corpus': train_corpus, 'relevant_docs': train_relevant_docs}, file, indent=4)
    
    with open(VAL_DATASET_PATH, 'w') as file:
        json.dump({'queries': val_queries, 'corpus': val_corpus, 'relevant_docs': val_relevant_docs}, file, indent=4)
    
    with open(TEST_DATASET_PATH, 'w') as file:
        json.dump({'queries': test_queries, 'corpus': test_corpus, 'relevant_docs': test_relevant_docs}, file, indent=4)

# Call the function with the training file name
split_data(DATASET)
```
