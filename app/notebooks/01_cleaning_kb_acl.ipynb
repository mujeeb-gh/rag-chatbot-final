```python
import pandas as pd
import re
```

### Load the Dataset (Knowledge Base)

```python
df = pd.read_parquet("../datasets/KB/acl-publication-info.74k.parquet")
```

### EDA to understand the dataset better

```python
df.info()
```

```python
df.nunique()
```

### Sort by year in Descending Order

```python
df.sort_values(by='year', ascending=False, inplace=True)
df
```

```python
df.drop_duplicates(inplace = True)
```

## Data Cleaning

### Drop some irrelevant columns (columns not needed for training RAG model)

```python
df = df.drop(columns=["corpus_paper_id", "pdf_hash", "numcitedby", "publisher", "address",
    "month", "booktitle", "pages", "number", "volume", "journal", "editor", "isbn", "abstract", "doi"])
df.info()
```

### Renaming Columns

```python
new_column_names = {
    "full_text": "text",  # Assuming full_text becomes the text column
    # "year": "publication_year",  # Example renaming year
    # "author": "authors",  # Assuming multiple authors (plural)
}

df.rename(columns=new_column_names, inplace=True)
df
```

### Text Cleaning - Cleans text data (lowercase, removes special characters)

```python
def clean_text(text):
    text = text.lower()
    text = re.sub(r"[^a-zA-Z0-9\s]", "", text)  # Remove non-alphanumeric characters and spaces
    return text

# Apply cleaning to relevant columns
df["text"] = df["text"].apply(clean_text)
df["title"] = df["title"].apply(clean_text)
```

### Drop columns with null values now

```python
df.dropna(inplace=True)
df
```

```python
df.to_csv("cleaned_kb_acl.csv", index=False)
```
